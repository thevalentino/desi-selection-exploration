# Expected flux for [OII]7320,7330

Rough order of magnitude to start, OII is 1% of OIII5007.


```python
from astropy.table import Table
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = Table.read("../data/emp-candidates-v261018.fits").to_pandas()
```

```python
plt.hist(np.log10(a['OIII5007_FLUX']) - 17, bins=np.linspace(-18, -14, 30))
plt.xlabel("log10 (OIII 5007 / [1e-17 erg/s/cm^2]")
```

```python
expected_oii7330 = 0.01 * a['OIII5007_FLUX']
plt.hist(np.log10(expected_oii7330) - 17, bins=np.linspace(-20, -16, 30))
plt.xlabel("OII 7330 flux [1e-17 erg/s/cm^2]")
```
